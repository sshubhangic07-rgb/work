```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt

# Load dataset
digits = load_digits()

print("Dataset shape:", digits.data.shape)
print("Image shape:", digits.images.shape)
print("Target labels:", digits.target_names)

# Display the 6th image (index 5)
plt.figure(figsize=(4, 4))
plt.imshow(digits.images[0], cmap='gray')
plt.title(f"Label: {digits.target[5]}")
plt.axis("off")
plt.show()
```

```python

import matplotlib.pyplot as plt
digits = load_digits()
fig,axes = plt.subplots(5,2,figsize=(10,5))
for i,ax in enumerate(axes.flat):
    ax.imshow(digits.images[i],cmap='gray')
    ax.set_title(f"digit:{digits.target[i]}")
    ax.axis('off')

plt.tight_layout()
plt.show()
```
